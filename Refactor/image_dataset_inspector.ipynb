```python
import cv2
import os
import glob

def inspect_image_dimensions(folder_path, valid_extensions=('*.jpg', '*.jpeg', '*.png', '*.bmp')):
    image_paths = []
    for ext in valid_extensions:
        image_paths.extend(glob.glob(os.path.join(folder_path, ext)))
    image_paths = sorted(image_paths)

    if not image_paths:
        print(f"هیچ تصاویری در مسیر '{folder_path}' یافت نشد.")
        return

    print("==========================================================================")
    print(f"📷 بررسی مشخصات و ابعاد عکس‌ها (تعداد کل: {len(image_paths)})")
    print("==========================================================================")
    print(f"{'نام فایل':<30} | {'ابعاد (Width x Height)':<22} | {'مگاپیکسل':<12} | {'حجم فایل':<10}")
    print("-" * 80)

    total_pixels_sum = 0

    for path in image_paths:
        # خواندن تصویر
        img = cv2.imread(path)
        if img is None:
            continue

        h, w = img.shape[:2]
        total_pixels = w * h
        megapixels = total_pixels / 1_000_000
        file_size_mb = os.path.getsize(path) / (1024 * 1024)

        total_pixels_sum += total_pixels
        filename = os.path.basename(path)

        print(f"{filename:<30} | {w} x {h:<16} | {megapixels:.2f} MP      | {file_size_mb:.2f} MB")

    avg_pixels = total_pixels_sum / len(image_paths) if image_paths else 0
    print("==========================================================================")
    print(f"📊 میانگین رزولوشن تصاویر: {avg_pixels / 1_000_000:.2f} مگاپیکسل")
    print("==========================================================================\n")

# اجرا روی همان مسیر پروژه شما
inspect_image_dimensions("C:/Users/win/Desktop/SunTrackerSplit/SunTracker-Refactor/Refactor/Images")
```
